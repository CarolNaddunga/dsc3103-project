### Lab 3

**Construct Source A + Source B → Validate → Clean → Transform → Store**. Running one documented command reproduces the processed dataset from raw sources(two sources) with no manual file editing.  This turns lab_2 profiler/cleaner into the first stage of a real pipeline.

**Step 1 — Create the setup below before adding a second data source.**  Notice that this now setsup a package with a single entry point:
```
src/
├── ingest/
│   ├── source_a.py  #reads data/raw/prices.csv
│   └── source_b.py  #calls an API(rainfall),or reads a second file(csv)/table
├── validate/
│   └── rules.py     #done already — reuse and extend, don't rewrite
├── transform/
│   └── clean.py     #done already — reuse and extend
├── common/
│   ├── config.py    #loads settings from a file/env (not hard-coded)
│   └── logging_setup.py
└── run_pipeline.py  #the ONE documented entry point
```

**Step 2 — Add a second source.** Pick a genuinely different source type from Week 2's CSV — an API call (e.g., a public rainfall API for your domain, or a synthetic API), a second file format (JSON), or a second table in a small local database. The two sources should plausibly join or align on something (e.g., market + date, or district + date) — Please make sure you use two somehow related datasets as this is important in the proceeding weeks particularly on storage/schema design.

**Step 3 — Decide ETL or ELT, and justify it.** Write 3–4 sentences in `docs/week3_notes.md`: are you transforming before or after loading, and why, given your two sources and your target store? There is no single correct answer — the point is a defensible choice, not a memorized rule.

**Step 4 — Make stage boundaries explicit.** Each stage (ingest, validate, transform, load) should be a function with a clear input/output contract — e.g., `ingest_source_a(path: str) -> pd.DataFrame`, `validate(ruleX(df: pd.DataFrame)) -> DataFrame` (failed rows, with a reasons), `transform/clean(all rulesX(df:pd.DataFrame) -> DataFrame, log:list[dict])`. Do not write one function called `pipeline()` that does everything inline — if you can't test a stage in isolation, it isn't a stage.

**Step 5 — Configuration, not hard-coding.** File paths, the API URL, and any thresholds (e.g., "reject prices below X") belong in a config file or environment variables, loaded once in `src/common/config.py` — never hard-coded inside `ingest/` or `transform/` functions.

**Step 6 — Fail fast, and say why.** Decide, per validation rule, whether a failure should stop the whole run (e.g., the source file is missing entirely) or be logged and the bad rows quarantined (e.g., a handful of invalid prices). Both are legitimate — the difference is `raise` vs. "log and continue," and it must be a deliberate choice per rule, not accidental.

**Step 7 — Logging.** Every run of `run_pipeline.py` should produce a log showing: which stage ran, how many rows went in/out of each stage, and any errors — not just `print()` statements scattered through the code. A groupmate should be able to read the log and know what happened without re-running anything.

**Step 8 — Make it idempotent.** Running `run_pipeline.py` twice in a row on the same input must not duplicate rows in your processed output. Decide and implement one mechanism: overwrite the output file/table each run, or upsert by a key, or check-and-skip if output already exists and is current. State which one you chose and why in `docs/week3_notes.md`.

**Step 9 — Write tests.** Using `pytest` (or similar), write tests for at least two of your pipeline functions — e.g., a validation rule correctly rejects a bad row, and the transform step correctly handles a missing value the way you documented. These are small, fast, no-infrastructure-needed tests, not end-to-end runs.

**Step 10 — Break it on purpose, then fix it.** Simulate a schema change (rename a column in source B, or change a date format) or a malformed record, and run the pipeline. It should fail *explicitly and informatively* — not crash with an unrelated stack trace, and not silently produce wrong output. Fix the pipeline so this failure mode is now handled (reject cleanly, or adapt), and capture both the failing run's log and the corrected run's log as evidence.

**Step 11 — Commit incrementally**, matching the steps above (refactor → second source → config → logging → idempotency → tests → failure fix). Six-plus explained commits, not one.
